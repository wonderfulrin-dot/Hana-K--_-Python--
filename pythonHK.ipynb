```python
# 두 숫자 입력받기
num1 = float(input("첫 번째 숫자를 입력하세요: "))
num2 = float(input("두 번째 숫자를 입력하세요: "))

# 연산자 입력받기
operator = input("연산자를 입력하세요 (+, -, *, /): ")

# 계산
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 == 0:
        result = "0으로 나눌 수 없습니다."
    else:
        result = num1 / num2
else:
    result = "올바른 연산자를 입력해주세요."

# 결과 출력
print("결과:", result)
```

```python
import yfinance as yf

# NVIDIA 주식 데이터 가져오기
nvda = yf.Ticker("NVDA")

data = nvda.history(start="2020-01-02", end="2020-01-03")

# 2020년 1월 2일 종가
buy_price = data["Close"].iloc[0]

# 현재 주가
current_data = nvda.history(period="1d")
current_price = current_data["Close"].iloc[-1]

# 투자금
investment = 10_000_000

# 달러 환율을 고려하지 않고 단순 주가 수익률 계산
shares = investment / buy_price
current_value = shares * current_price
profit = current_value - investment
return_rate = (profit / investment) * 100

print("2020년 매수가:", buy_price)
print("현재 주가:", current_price)
print("보유 주식 수:", shares)
print("현재 평가금액:", current_value, "원")
print("수익금:", profit, "원")
print("수익률:", return_rate, "%")
```
